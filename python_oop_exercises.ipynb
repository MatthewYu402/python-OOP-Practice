### Problem 3: Bank Account Class

Create a `BankAccount` class with:
- Attributes: `account_number`, `balance`, `owner_name`
- Methods: `deposit(amount)`, `withdraw(amount)`, `get_balance()`
- Ensure withdrawals don't allow negative balances


```python
# Exercise 3: Bank Account Class
# Write your code here
class BankAccount():
    def __init__(self, account_number, balance, owner_name):
        self.account_number = account_number
        self.balance = balance
        self.owner_name = owner_name
        
    def deposit(self, amount):
        if amount < 0:
            print("Cannot deposit negative balance")
            return
        self.balance += amount
        
        
    def withdraw(self, amount):
        if self.balance - amount < 0:
            print(f"You have ${self.balance}. Cannot allow negative balance from withdrawing ${amount}")
        self.balance -= amount
        
    def get_balance(self):
        return self.balance
    
    
# Test your code
account = BankAccount("12345", 1000, "John Doe")
account.deposit(500)
account.withdraw(200)
print(account.get_balance())  # Expected: 1300
account.withdraw(2000)  # Should not allow negative balance
```

### Problem 4: Rectangle Class

Create a `Rectangle` class with:
- Attributes: `width`, `height`
- Methods: `area()`, `perimeter()`
- A method `is_square()` that returns True if width equals height


```python
# Exercise 4: Rectangle Class
# Write your code here
class Rectangle():
    def __init__(self, width, height):
        self.width = width
        self.height = height
        
    def area(self):
        return self.width * self.height
    
    def perimeter(self):
        return (2 * self.width) + (2 * self.height)
    
    def is_square(self):
        if self.width == self.height:
            return True
        
        return False
    
        

# Test your code
rect = Rectangle(5, 5)
print(rect.area())  # Expected: 25
print(rect.perimeter())  # Expected: 20
print(rect.is_square())  # Expected: True
```

### Problem 5: Student Class

Create a `Student` class with:
- Attributes: `name`, `student_id`, `grades` (list)
- Methods: `add_grade(grade)`, `get_average()`, `get_letter_grade()`
- Letter grades: A (>=90), B (>=80), C (>=70), D (>=60), F (<60)


```python
# Exercise 5: Student Class
# Write your code here
class Student():
    def __init__(self, name, student_id):
        self.name = name
        self.student_id = student_id
        self.grades = []
        
    def add_grade(self, grade):
        self.grades.append(grade)
        
    def get_average(self):
        return sum(self.grades) / len(self.grades)
    
    def get_letter_grade(self):
        avg = self.get_average()
        
        match avg // 10:
            case 10:
                return 'A'
            case 9:
                return 'A'
            case 8:
                return 'B'
            case 7:
                return 'C'
            case 6:
                return 'D'
            case _:
                return 'F'

# Test your code
student = Student("Alice", "S001")
student.add_grade(85)
student.add_grade(92)
student.add_grade(78)
print(student.get_average())  # Expected: 85.0
print(student.get_letter_grade())  # Expected: 'B'
```

### Problem 8: Calculator Class

Create a `Calculator` class with:
- An attribute `history` (list to store operations)
- Methods: `add(a, b)`, `subtract(a, b)`, `multiply(a, b)`, `divide(a, b)`
- Each method should store the operation in history
- A method `get_history()` to return all operations


```python
# Exercise 8: Calculator Class
# Write your code here
class Calculator():
    def __init__(self):
        self.history = []
        
    def add(self, a, b):
        result = a + b
        self.history.append(('add', a, b, result))
        
    def subtract(self, a, b):
        result = a - b
        self.history.append(('subtract', a, b, result))
        
    def multiply(self, a, b):
        result = a * b
        self.history.append(('multiply', a, b, result))
        
    def divide(self, a, b):
        result = a / b
        self.history.append(('divide', a, b , result))
        
    def get_history(self):
        return self.history

# Test your code
calc = Calculator()
calc.add(5, 3)
calc.multiply(4, 2)
print(calc.get_history())  # Expected: [('add', 5, 3, 8), ('multiply', 4, 2, 8)]
```

### Problem 10: Shopping Cart Class

Create a `ShoppingCart` class with:
- An attribute `items` (dictionary: item_name -> quantity)
- Methods: `add_item(item, quantity)`, `remove_item(item)`, `get_total_items()`, `clear_cart()`


```python
# Exercise 10: Shopping Cart Class
# Write your code here
class ShoppingCart():
    def __init__(self):
        self.items = []
        
    def add_item(self, item, quantity):
        self.items.append((item, quantity))
        
    def remove_item(self, item):
        for thing in self.items:
            if thing[0] == item:
                self.items.remove(thing)
        
    def get_total_items(self):
        total = 0
        for item in self.items:
            total += item[1]
            
        return total
            
    def clear_cart(self):
        self.items.clear()
        


# Test your code
cart = ShoppingCart()
cart.add_item("apple", 5)
cart.add_item("banana", 3)
print(cart.get_total_items())  # Expected: 8
cart.remove_item("apple")
print(cart.get_total_items())  # Expected: 3
```

### Problem 14: Timer Class

Create a `Timer` class with:
- Attributes: `start_time`, `end_time`
- Methods: `start()`, `stop()`, `elapsed_time()` (returns time difference in seconds)
- Use `time.time()` from the time module


```python
# Exercise 14: Timer Class
import time
# Write your code here
class Timer():
    def __init__(self):
        self._start_time = None
        self._elapsed_time = 0.0
        
    def start(self):
        if self.start_time is None:
            raise RuntimeError("Timer is already running. Use stop() before starting again")
        self._start_time = time.perf_counter()
        
    def stop(self):
        if self.start_time is None:
            raise RuntimeError("Timer is not running. Use start() to begin")
        end_time = time.perf_counter()
        print(f"Timer stopped")
        
    def elapsed_time(self):
        self._elasped_time += end_time - self._start_time
        self._start_time = None

# Test your code
# timer = Timer()
# timer.start()
# time.sleep(1)  # Simulate 1 second passing
# timer.stop()
# print(timer.elapsed_time())  # Expected: ~1.0 seconds
```

### Problem 15: Fraction Class

Create a `Fraction` class with:
- Attributes: `numerator`, `denominator`
- Methods: `add(other)`, `multiply(other)`, `simplify()` (reduce to lowest terms)
- Handle division by zero in the denominator


```python
# Exercise 15: Fraction Class
import math
# Write your code here
class Fraction():
    def __init__(self, numerator, denominator):
        self.numerator = numerator
        self.denominator = denominator
        
    def add(self, other):
        self.numerator += other.numerator
        self.denominator += other.numerator
        return self
        
    def multiply(self, other):
        self.numerator *= other.numerator
        self.denominator *= other.numerator
        return self
        
    def simplify():
        if self.denominator == 0:
            print(f"Divide by zero error!")
            return
        return self.numerator // self.denominator

# Test your code
frac1 = Fraction(3, 4)
frac2 = Fraction(1, 2)
result = frac1.add(frac2)
print(result.numerator, result.denominator)  # Expected: 5, 4
```

### Problem 17: Property Decorator

Create a `Circle` class with:
- A private attribute `_radius`
- Use `@property` and `@radius.setter` to control access to radius
- Ensure radius cannot be negative


```python
# Exercise 17: Property Decorator
# Write your code here
class Circle:
  def __init__(self, radius):
    self._radius = radius
    
  @property
  def radius(self):
    return self._radius
  
  @radius.setter
  def radius(self, value):
    if value < 0:
      raise ValueError("Radius cannot be negative")
    self._radius = value

# Test your code
circle = Circle(5)
print(circle.radius)  # Expected: 5
circle.radius = 1  # Should raise ValueError or set to 0
print(circle.radius)  # Expected: 1
```
