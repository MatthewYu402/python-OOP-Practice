### Problem 4: Rectangle Class

Create a `Rectangle` class with:
- Attributes: `width`, `height`
- Methods: `area()`, `perimeter()`
- A method `is_square()` that returns True if width equals height


```python
# Exercise 4: Rectangle Class
# Write your code here
class Rectangle():
    def __init__(self, width, height):
        self.width = width
        self.height = height
        
    def area(self):
        return self.width * self.height
    
    def perimeter(self):
        return (2 * self.width) + (2 * self.height)
    
    def is_square(self):
        if self.width == self.height:
            return True
        
        return False
    
        

# Test your code
rect = Rectangle(5, 5)
print(rect.area())  # Expected: 25
print(rect.perimeter())  # Expected: 20
print(rect.is_square())  # Expected: True
```

### Problem 14: Timer Class

Create a `Timer` class with:
- Attributes: `start_time`, `end_time`
- Methods: `start()`, `stop()`, `elapsed_time()` (returns time difference in seconds)
- Use `time.time()` from the time module


```python
# Exercise 14: Timer Class
import time
# Write your code here
class Timer():
    def __init__(self):
        self.start_time = None
        self.end_time = None
        
    def start(self):
        self.start_time = time.time()
        
    def stop(self):
        self.end_time = time.time()
        
    def elapsed_time(self):
        if self.start_time is None:
            return 0.0
        end = self.end_time if self.end_time else time.time()
        return end - self.start_time

# Test your code
timer = Timer()
timer.start()
time.sleep(1)  # Simulate 1 second passing
timer.stop()
print(timer.elapsed_time())  # Expected: ~1.0 seconds
```

### Problem 15: Fraction Class

Create a `Fraction` class with:
- Attributes: `numerator`, `denominator`
- Methods: `add(other)`, `multiply(other)`, `simplify()` (reduce to lowest terms)
- Handle division by zero in the denominator


```python
# Exercise 15: Fraction Class
import math
# Write your code here
class Fraction():
    def __init__(self, numerator, denominator):
        self.numerator = numerator
        self.denominator = denominator
        
    def add(self, other):
        self.numerator += other.numerator
        self.denominator += other.numerator
        return self
        
    def multiply(self, other):
        self.numerator *= other.numerator
        self.denominator *= other.numerator
        return self
        
    def simplify():
        if self.denominator == 0:
            print(f"Divide by zero error!")
            return
        return self.numerator // self.denominator

# Test your code
frac1 = Fraction(3, 4)
frac2 = Fraction(1, 2)
result = frac1.add(frac2)
print(result.numerator, result.denominator)  # Expected: 5, 4
```

### Problem 17: Property Decorator

Create a `Circle` class with:
- A private attribute `_radius`
- Use `@property` and `@radius.setter` to control access to radius
- Ensure radius cannot be negative


```python
# Exercise 17: Property Decorator
# Write your code here
class Circle:
  def __init__(self, radius):
        self._radius = radius
    
  @property
  def radius(self):
      return self._radius
  
  @radius.setter
  def radius(self, value):
      if value < 0:
          raise ValueError("Radius cannot be negative")
      self._radius = value
  
  @property
  def area(self):
      import math
      return math.pi * self._radius ** 2

# Test your code
circle = Circle(5)
print(circle.radius)  # Expected: 5
circle.radius = 1  # Should raise ValueError or set to 0
print(circle.radius)  # Expected: 1
```

```python
# Exercise 55: Complete Special Methods
# Write your code here
import math

class Fraction:
  def __init__(self, numerator, denominator):
    if denominator == 0:
      raise ValueError("Denominator cannot be zero")
    self.numerator = numerator
    self.denominator = denominator
    self.simplify()
    
  def simplify(self):
    gcd = math.gcd(self.numerator, self.denominator)
    self.numerator //= gcd
    self.denominator //= gcd
    
  def __str__(self):
    return f"{self.numerator}/{self.denominator}"
  
  def __repr__(self):
    return f"Fraction({self.numerator}/ {self.denominator})"
  
  def __add__(self, other):
    new_num = self.numerator * other.denominator + other.numerator * self.denominator
    new_den = self.denominator * other.denominator
    return Fraction(new_num, new_den)

  def __sub__(self, other):
    new_num = self.numerator * other.denominator - other.numerator * self.denominator
    new_den = self.denominator * other.denominator
    return Fraction(new_num, new_den)

  def __mul__(self, other):
    return Fraction(self.numerator * other.numerator, self.denominator * other.denominator)

  def __eq__(self, other):
    return self.numerator == other.numerator and self.denominator == other.denominator

  def __lt__(self, other):
    return self.numerator * other.denominator < other.numerator * self.denominator
      

# Test your code
f1 = Fraction(1, 2)
f2 = Fraction(1, 4)
print(f1 + f2)  # Expected: Fraction(3, 4)
print(f1 < f2)  # Expected: False
```

### Problem 58: Static Method - Utility Functions

Create a `MathUtils` class with static methods:
- `is_prime(n)`, `factorial(n)`, `gcd(a, b)`


```python
# Exercise 58: Static Method - Utility Functions
# Write your code here
class MathUtils:
  @staticmethod
  def is_prime(n):
    if n < 2:
      return False
    for i in range (2, int(n**0.5) + 1):
      if n % i == 0:
        return False
    return True
  
  @staticmethod
  def factorial(n):
    if n < 0:
      raise ValueError("Factorial not defined for negative numbers")
    result = 1
    for i in range(1, n + 1):
      result *= i
    return result

  
  @staticmethod
  def gcd(a, b):
    while b:
      a, b = b, a % b
    return a
  


# Test your code
print(MathUtils.is_prime(7))  # Expected: True
print(MathUtils.factorial(5))  # Expected: 120
print(MathUtils.gcd(48, 18))  # Expected: 6
```

### Problem 62: Class Method Chain

Create a `QueryBuilder` class with class methods:
- `select(fields)`, `from_table(table)`, `where(condition)`
- Each should return `self` to allow method chaining


```python
# Exercise 62: Class Method Chain
# Write your code here
class QueryBuilder:
  def __init__(self):
    self.select_fields = None
    self.from_table_name = None
    self.where_clause = None
    
  @classmethod
  def select(cls, fields):
    instance = cls()
    instance.select_fields = fields
    return instance
  
  def from_table(self, from_table):
    self.from_table_name = from_table
    return self
  
  def where(self, condition):
    self.where_clause = condition
    return self
  
  def build(self):
    query = f"SELECT {self.select_fields} FROM {self.from_table_name}"
    if self.where_clause:
      query += f" WHERE {self.where_clause}"
      
    return query
      

# Test your code
query = QueryBuilder.select("name, email").from_table("users").where("age > 18")
print(query.build())  # Should return SQL query string
```
